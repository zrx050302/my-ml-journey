## w1 d3 代码练习

```python
#设计数据
import numpy as np 
a=np.random.randint(0,100,5)
print(a)
b=np.random.randint(0,100,5)
print(b)
```

```python
import matplotlib.pyplot as plt


def g(x,w,c):
    y=w*x+c
    return y
def J(b1,b):
    b2=(b1-b)**2
    s=1/2/len(a)*np.sum(b2)
    return s
def r(w,al,b1,a,b,c):
    m=np.abs(al*1/len(a)*np.sum((b1-b)*a))
    n=np.abs(al*1/len(a)*np.sum(b1-b))
    w=w-al*1/len(a)*np.sum((b1-b)*a)
    c=c-al*1/len(a)*np.sum(b1-b)
    return w,c,m,n
#shape返回的是元组
w=0
c=0
al=0.00001
i=1
j=1
h=0
u=[]
#i,j自己忘了
while (i>=0.001 or j>=0.001) and h<100000:#  最大迭代次数
    b1=g(a,w,c)
    w,c,i,j=r(w,al,b1,a,b,c)
    h=h+1
    u.append(J(b1,b))

print(w,c,u[-1])
plt.plot(a,b1,'b-',label='predict')
plt.scatter(a,b,c='red',label='real')
#绘图技巧
plt.xlabel('X')
plt.ylabel('Y')
plt.title('曲线与散点绘制在同一图中')
plt.legend()        # 显示图例
plt.grid(True)      # 显示网格
plt.show()
#同一个字母涉及到多次使用
```

```python
u=np.array(u)

plt.plot(u,'b-',label='J')
plt.xlabel('X')
plt.ylabel('Y')

plt.legend()        # 显示图例
plt.grid(True)      # 显示网格
plt.show()#学习率过大和过小的问题
```
