```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 快速自测：用 NumPy 算一个梯度下降的最小雏形
w = 0.0
x = np.array([1, 2, 3, 4])
y = np.array([2, 4, 6, 8])   # y = 2x

for i in range(100):
    pred = w * x
    grad = np.mean((pred - y) * x)   # dJ/dw
    w = w - 0.01 * grad

print(f"学到的 w = {w:.4f}") 
```

```python
plt.plot(x, y, 'o')
plt.title("Environment Test")
plt.show()
```
